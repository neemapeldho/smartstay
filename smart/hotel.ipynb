```python
import pandas
import numpy
```

```python
data = pandas.read_csv("Airbnb_Data.csv")
df = pandas.DataFrame(data)
df
```

```python
df.columns
```

```python
df = df.drop(columns=["id","amenities","description","first_review","last_review","latitude","longitude","thumbnail_url"])
df
```

```python
df.isna().sum()
```

```python
df.dtypes
```

```python

df["host_has_profile_pic"] = df["host_has_profile_pic"].fillna(df["host_has_profile_pic"].mode()[0])
df["host_identity_verified"] = df["host_identity_verified"].fillna(df["host_identity_verified"].mode()[0])
df["host_response_rate"] = df["host_response_rate"].fillna(df["host_response_rate"].mode()[0])

from datetime import datetime

df['host_since'] = pandas.to_datetime(df['host_since'])
df['host_experience'] = (datetime.now() - df['host_since']).dt.days // 365
df['host_experience'] = df['host_experience'].fillna(df['host_experience'].median())
df = df.drop('host_since', axis=1) 

cat_cols = ['neighbourhood', 'zipcode']
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

num_cols = ['bathrooms', 'bedrooms', 'beds', 'review_scores_rating']
for col in num_cols:
    df[col] = df[col].fillna(df[col].mean())
```

```python
df.isna().sum()
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

for i in df.columns:
    if df[i].dtype == "object":
        df[i] = le.fit_transform(df[i].astype(str))
```

```python
df.dtypes
```

```python
# now spliting the data into input and output variables
y = df["log_price"]
X = df.drop(columns=["log_price"], axis=1)
X
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()
lr = lr.fit(X_train, y_train)
y_pred = lr.predict(X_test)
y_pred
```

```python
y_test
```

```python
print(lr.coef_)
print(lr.intercept_)
```

```python
from sklearn.metrics import mean_absolute_percentage_error
print(mean_absolute_percentage_error(y_test, y_pred))
```

```python
from sklearn.metrics import r2_score
print(r2_score(y_test, y_pred))
```
